```python
import re
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files
```

```python
up = files.upload()
nama_file = list(up.keys())[0]
print("File terbaca:", nama_file)
```

```python
JUMLAH_KOLOM_DEMOGRAFI = 6
df = pd.read_excel(nama_file)
butir = df.iloc[:, JUMLAH_KOLOM_DEMOGRAFI:]

kolom_durasi = [k for k in butir.columns if str(k).strip().startswith("Durasi")]
butir = butir.drop(columns=kolom_durasi)

print("Jumlah responden :", len(df))
print("Jumlah butir     :", butir.shape[1])
print("Data kosong      :", int(butir.isna().sum().sum()))
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5,
}
num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")

print("Sel yang gagal dikonversi:", int(num.isna().sum().sum()))
num.head()
```

```python
nama_rapi = {
    "Penggunaan AI pada tugas atau projek kuliah": "Penggunaan AI",
    "Aspek Psikologis penggunaan AI pada tugas atau projek kuliah": "Aspek Psikologis",
    "Kebiasaan penggunaan teknologi digital": "Kebiasaan Digital",
    "Prokrastinasi akademik.": "Prokrastinasi Akademik",
    "Rsspon emosional": "Respon Emosional",
}

dimensi = {}
for kolom in butir.columns:
    nama = re.match(r"^(.*?) \[", kolom).group(1).strip()
    nama = nama_rapi.get(nama, nama)
    dimensi.setdefault(nama, []).append(kolom)

for nama, daftar in dimensi.items():
    print(f"{nama}: {len(daftar)} butir")
```

```python
n = len(num)
t = stats.t.ppf(0.975, n - 2)
r_tabel = t / np.sqrt((n - 2) + t**2)
print(f"n = {n} | r tabel (α=0,05; df={n-2}) = {r_tabel:.3f}")

hasil_validitas = []
for nama, daftar in dimensi.items():
    x = num[daftar]
    total = x.sum(axis=1)
    for i, kolom in enumerate(daftar, 1):
        r, p = stats.pearsonr(x[kolom], total)
        r_koreksi, _ = stats.pearsonr(x[kolom], total - x[kolom])
        hasil_validitas.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Corrected r": round(r_koreksi, 3),
            "p-value": round(p, 4),
            "Keterangan": "Valid" if r > r_tabel else "Tidak valid",
        })

tabel_validitas = pd.DataFrame(hasil_validitas)
tabel_validitas
```

```python
def cronbach_alpha(x):
    k = x.shape[1]
    return k / (k - 1) * (1 - x.var(ddof=1).sum() / x.sum(axis=1).var(ddof=1))

def kategori(a):
    return ("Sangat baik" if a >= 0.90 else "Baik" if a >= 0.80
            else "Reliabel" if a >= 0.70 else "Tidak reliabel")

hasil_reliabilitas = []
alpha_dihapus = []
for nama, daftar in dimensi.items():
    x = num[daftar]
    a = cronbach_alpha(x)
    hasil_reliabilitas.append({
        "Dimensi": nama,
        "Jumlah butir": len(daftar),
        "Cronbach's alpha": round(a, 3),
        "Kategori": kategori(a),
    })
    for i, kolom in enumerate(daftar, 1):
        if len(daftar) > 2:
            alpha_dihapus.append({
                "Dimensi": nama,
                "Butir": f"B{i}",
                "Alpha jika dihapus": round(cronbach_alpha(x.drop(columns=kolom)), 3),
            })

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)
display(tabel_reliabilitas)
display(tabel_alpha_dihapus)
```

```python
gabung = tabel_validitas.merge(tabel_alpha_dihapus, on=["Dimensi", "Butir"], how="left")

with pd.ExcelWriter("hasil_uji_validitas_reliabilitas.xlsx") as w:
    gabung.to_excel(w, sheet_name="Validitas", index=False)
    tabel_reliabilitas.to_excel(w, sheet_name="Reliabilitas", index=False)

files.download("hasil_uji_validitas_reliabilitas.xlsx")
gabung
```

```python
print("Butir tidak valid (r hitung <= r tabel):")
print(tabel_validitas[tabel_validitas["Keterangan"] == "Tidak valid"])
print()
print("Valid menurut r hitung > r tabel :", int((tabel_validitas["r hitung"] > r_tabel).sum()))
print("Valid menurut corrected r >= 0,30:", int((tabel_validitas["Corrected r"] >= 0.30).sum()))

# Butir yang alpha-jika-dihapusnya lebih tinggi daripada alpha dimensinya
alpha_dim = tabel_reliabilitas.set_index("Dimensi")["Cronbach's alpha"]
cek = gabung.copy()
cek["Alpha dimensi"] = cek["Dimensi"].map(alpha_dim)
print()
print("Butir dengan alpha jika dihapus > alpha dimensi:")
cek[cek["Alpha jika dihapus"] > cek["Alpha dimensi"]][
    ["Dimensi", "Butir", "r hitung", "Corrected r", "Alpha jika dihapus", "Alpha dimensi"]]
```

```python
dim_uji = "Penggunaan AI"
butir_buang = 4  # B4 (butir ke-4) pada dimensi tersebut

daftar = dimensi[dim_uji]
x_awal = num[daftar]
x_baru = x_awal.drop(columns=daftar[butir_buang - 1])

print(f"Alpha {dim_uji} sebelum : {cronbach_alpha(x_awal):.3f}")
print(f"Alpha {dim_uji} sesudah : {cronbach_alpha(x_baru):.3f}")

total_baru = x_baru.sum(axis=1)
ulang = []
for kolom in x_baru.columns:
    nomor = daftar.index(kolom) + 1  # nomor butir asli (B4 tidak ada)
    r, p = stats.pearsonr(x_baru[kolom], total_baru)
    rk, _ = stats.pearsonr(x_baru[kolom], total_baru - x_baru[kolom])
    ulang.append({"Butir": f"B{nomor}", "r hitung": round(r, 3),
                  "Corrected r": round(rk, 3), "Valid": r > r_tabel})
pd.DataFrame(ulang)
```

```python
def validitas_subset(data):
    n_s = len(data)
    ts = stats.t.ppf(0.975, n_s - 2)
    rt = ts / np.sqrt((n_s - 2) + ts**2)
    r_all = []
    for nama, daftar in dimensi.items():
        x = data[daftar]
        total = x.sum(axis=1)
        for i, kolom in enumerate(daftar, 1):
            r, _ = stats.pearsonr(x[kolom], total)
            r_all.append({"Dimensi": nama, "Butir": f"B{i}", "r": r})
    hasil = pd.DataFrame(r_all)
    hasil["valid"] = hasil["r"] > rt
    return rt, hasil

baris = []
for ukuran in [10, 15, 20, 25, 30, len(num)]:
    data = num if ukuran == len(num) else num.sample(ukuran, random_state=1)
    rt, hasil = validitas_subset(data)
    tidak = hasil[~hasil["valid"]]
    baris.append({
        "n": ukuran,
        "r tabel": round(rt, 3),
        "Butir valid": int(hasil["valid"].sum()),
        "Butir tidak valid": ", ".join(f"{d} {b}" for d, b in zip(tidak["Dimensi"], tidak["Butir"])) or "-",
    })
pd.DataFrame(baris)
```
